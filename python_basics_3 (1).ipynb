## 🛠 Утилита для автопроверок

Следующая ячейка определяет вспомогательные функции. **Запустите её один раз перед началом практики.**

````python
# Утилиты для автоматической проверки заданий.
# Запустите эту ячейку один раз.

import io
import contextlib


def check_value(name, actual, expected):
    """Проверяет, что переменная равна ожидаемому значению."""
    if actual == expected:
        print(f"✅ Задание пройдено: {name} = {actual!r}")
        return True
    else:
        print(f"❌ Ошибка: ожидалось {name} = {expected!r}, получено {actual!r}")
        return False


def check_output(student_func, expected_output):
    """Запускает student_func(), перехватывает stdout и сравнивает с ожидаемым."""
    buffer = io.StringIO()
    try:
        with contextlib.redirect_stdout(buffer):
            student_func()
    except Exception as e:
        print(f"❌ Ошибка при выполнении: {type(e).__name__}: {e}")
        return False

    actual = buffer.getvalue().strip()
    expected = expected_output.strip()

    if actual == expected:
        print("✅ Задание пройдено")
        return True
    else:
        print("❌ Вывод не совпадает с ожидаемым.")
        print("--- Ожидалось ---")
        print(expected)
        print("--- Получено ---")
        print(actual)
        return False


print("Утилиты автопроверки загружены.")
````

# Часть 1. Строки

## 1. Что такое строка

**Строка** (`str`) — это неизменяемая последовательность символов. Строки используются для хранения текста: слов, предложений, целых абзацев.

### Создание строк

    s1 = 'hello'
    s2 = "world"
    s3 = '''многострочная
    строка'''

Строки можно складывать (конкатенация) и умножать на число:

    'ab' + 'cd'   # 'abcd'
    'ab' * 3      # 'ababab'

### Длина строки

    len('hello')  # 5

### Индексы

К символам строки можно обращаться по индексу. Индексы начинаются с **0**. Отрицательные индексы отсчитываются с конца:

| Индекс | 0 | 1 | 2 | 3 | 4 |
|--------|---|---|---|---|---|
| Символ | h | e | l | l | o |
| С конца | -5 | -4 | -3 | -2 | -1 |

    s = 'hello'
    s[0]    # 'h'
    s[-1]   # 'o'

**Важно:** строка **неизменяема**. Нельзя написать `s[0] = 'H'` — будет ошибка. Нужно создавать новую строку.

````python
# Пример 1. Базовые операции со строками
s = "hello"

print("Строка:", s)
print("Длина:", len(s))
print("Первый символ:", s[0])
print("Последний символ:", s[-1])
print("Конкатенация:", s + " world")
print("Повтор:", "ab" * 3)
````

## 2. Срезы строк

**Срез** — это подстрока, полученная по правилу `s[start:stop:step]`.

- `start` — индекс начала (включается);
- `stop` — индекс конца (**не включается**);
- `step` — шаг (по умолчанию 1).

Примеры:

    s = 'python'
    s[1:4]    # 'yth'
    s[:3]     # 'pyt'
    s[3:]     # 'hon'
    s[::2]    # 'pto'
    s[::-1]   # 'nohtyp'  ← разворот строки

Срезы **не вызывают ошибку**, если выходят за границы — просто возвращают то, что есть.

````python
# Пример 2. Срезы
s = "python"

print("s[1:4] =", s[1:4])
print("s[:3]  =", s[:3])
print("s[3:]  =", s[3:])
print("s[::2] =", s[::2])
print("s[::-1]=", s[::-1])
print(("s[0:9]=", s[0:9]))
````

## 3. Методы строк

У строк много полезных методов. Они вызываются через точку и **не изменяют** исходную строку — возвращают новую.

| Метод | Что делает |
|-------|------------|
| `s.lower()` | Нижний регистр |
| `s.upper()` | Верхний регистр |
| `s.strip()` | Убирает пробелы и `\n` по краям |
| `s.replace(a, b)` | Заменяет все `a` на `b` |
| `s.split(sep)` | Разбивает строку по разделителю |
| `s.join(seq)` | Склеивает элементы через разделитель |
| `s.startswith(p)` | Начинается ли с `p`? |
| `s.endswith(p)` | Заканчивается ли на `p`? |
| `s.find(sub)` | Индекс первого вхождения или `-1` |
| `s.count(sub)` | Сколько раз встречается |
| `s.isdigit()` | Все ли символы — цифры? |
| `s.isalpha()` | Все ли символы — буквы? |

### Пример

    text = "  Hello, World!  "
    text.strip()               # 'Hello, World!'
    text.strip().lower()       # 'hello, world!'
    text.strip().split(', ')   # ['Hello', 'World!']

````python
# Пример 3. Методы строк
text = "  Hello, World!  "

print("strip:", text.strip())
print("lower:", text.strip().lower())
print("upper:", text.strip().upper())
print("replace:", text.replace("World", "Python"))
print("split:", "apple,banana,cherry".split(","))
print("join:", "-".join(["a", "b", "c"]))
print("startswith:", "python".startswith("py"))
print("count:", "banana".count("a"))
````

# Часть 2. Списки

## 5. Что такое список

**Список** (`list`) — упорядоченная изменяемая коллекция элементов. В отличие от строки, список можно менять: добавлять, удалять, заменять элементы.

### Создание списка

    words = ["AI", "tools", "help", "learn"]
    numbers = [1, 2, 3, 4, 5]
    mixed = [1, "two", 3.0, True]
    empty = []

### Индексы и срезы

Работают так же, как у строк:

    words[0]      # 'AI'
    words[-1]     # 'learn'
    words[1:3]    # ['tools', 'help']

### Длина списка

    len(words)    # 4

### Изменение элемента

    words[0] = "Python"
    print(words)   # ['Python', 'tools', 'help', 'learn']

````python
# Пример 4. Базовые операции со списками
words = ["AI", "tools", "help", "learn"]

print("Список:", words)
print("Длина:", len(words))
print("Первый:", words[0])
print("Последний:", words[-1])
print("Срез:", words[1:3])

words[0] = "Python"
print("После замены:", words)
````

## 6. Методы списков

Списки в отличие от строк **изменяемы**: методы меняют сам список.

| Метод | Что делает |
|-------|------------|
| `lst.append(x)` | Добавляет `x` в конец |
| `lst.extend(seq)` | Добавляет все элементы из `seq` |
| `lst.insert(i, x)` | Вставляет `x` перед индексом `i` |
| `lst.remove(x)` | Удаляет первое вхождение `x` |
| `lst.pop(i)` | Удаляет и возвращает элемент по индексу `i` (по умолчанию последний) |
| `lst.sort()` | Сортирует список (меняет сам список) |
| `lst.reverse()` | Разворачивает список |
| `lst.index(x)` | Индекс первого вхождения `x` |
| `lst.count(x)` | Сколько раз встречается `x` |

### Разница между `append` и `extend`

    a = [1, 2, 3]
    a.append([4, 5])       # [1, 2, 3, [4, 5]]  ← один элемент-список
    b = [1, 2, 3]
    b.extend([4, 5])       # [1, 2, 3, 4, 5]    ← три отдельных элемента

````python
# Пример 5. Методы списков
words = ["AI", "tools", "help"]

words.append("learn")
print("append:", words)

words.extend(["English", "teachers"])
print("extend:", words)

words.insert(0, "Python")
print("insert:", words)

words.remove("help")
print("remove:", words)

popped = words.pop()
print("pop вернул:", popped)
print("После pop:", words)

words.sort()
print("sort:", words)

words.reverse()
print("reverse:", words)
````

# Часть 3. Перебор и генераторы списков

## 7. Перебор строк и списков в цикле

    for word in ["AI", "tools", "help"]:
        print(word.upper())

Если нужен индекс, используйте `enumerate`:

    for i, word in enumerate(["AI", "tools", "help"]):
        print(i, word)

Чтобы получить **новый** список, применяя операцию к каждому элементу, используйте **генератор списков** (list comprehension):

    [w.upper() for w in ["AI", "tools"]]   # ['AI', 'TOOLS']
    [w for w in ["AI", "tools"] if len(w) > 2]  # ['tools']

**Осторожно:** не изменяйте список во время перебора. Если нужно удалять — перебирайте копию: `for x in lst[:]:`.

````python
# Пример 6. Перебор и генераторы списков
words = ["AI", "tools", "help", "learn"]

print("Простой перебор:")
for w in words:
    print(" ", w.upper())

print("С индексом:")
for i, w in enumerate(words):
    print(" ", i, w)

print("Генератор:", [w.upper() for w in words])
print("Фильтр по длине:", [w for w in words if len(w) > 3])
````

# Часть 4. Типичные ошибки

| Ошибка | Что происходит | Как исправить |
|--------|----------------|----------------|
| `s[100]` для строки короче 100 | `IndexError` | Проверяйте `len(s)`; используйте срез — он безопасен |
| `s[0] = 'H'` | `TypeError`: строка неизменяема | Создайте новую строку: `'H' + s[1:]` |
| `lst[100]` | `IndexError` | Проверяйте `len(lst)` |
| `lst.append([1,2])` вместо `extend` | Список становится вложенным | Используйте `extend`, если нужно добавить элементы по отдельности |
| Изменение списка внутри `for` | Пропуск элементов | Перебирайте копию: `for x in lst[:]:` |
| Забыли скобки у метода: `s.lower` | Возвращается объект-метод, а не строка | Вызывайте: `s.lower()` |
| `sort()` возвращает `None` | `x = lst.sort()` → `x is None` | Используйте `sorted(lst)` или вызывайте `lst.sort()` без присваивания |
| `'a' in 'abc'` vs `'a' in ['abc']` | Разные типы проверок | Помните: `in` для строки — подстрока, для списка — элемент |
